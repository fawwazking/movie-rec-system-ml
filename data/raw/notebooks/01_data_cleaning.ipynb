```python
import pandas as pd
import numpy as np
df = pd.read_csv('../src/letterboxd_movies_dataset.csv')
print(f"dataset Letterboxd Movies: {df.shape[0]} film")
print(f"Kolom ready: {df.columns.tolist()}")
print("\n5 first movies:")
display(df.head())
```

```python

print("Status Data Letterboxd:")
print("=" * 60)
print(f"\ntotal film: {len(df):,}")
print(f"total colom: {len(df.columns)}")
print("\nMissing Values per kolom:")
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
if len(missing) > 0:
    print(missing)
else:
    print("clean")

print("\nInfo main colom:")
print(f"  - title: {df['title'].nunique()} film")
print(f"  - yeae: {df['year'].min():.0f} - {df['year'].max():.0f}")
print(f"  - genre: {df['genres'].nunique()} combine genre")
print(f"  - runtime: {df['runtime'].min():.0f} - {df['runtime'].max():.0f} minute")
```

```python

print("duplicata data check:")
duplicate_titles = df[df.duplicated(subset=['title'], keep=False)]
print(f"\nduplicate on title: {len(duplicate_titles) // 2} ")
if len(duplicate_titles) > 0:
    print("\nFilm with duplicate title:")
    print(duplicate_titles[['title', 'year', 'genres']].head())
    df = df.drop_duplicates(subset=['title', 'year'], keep='first')
    print(f"\nclear: {len(df)} film")
else:
    print("no movies dupliacate")

print(f"\n✓ Data letterbox ready to use: {len(df):,} film")
```

```python
import os
os.makedirs('../processed', exist_ok=True)
df.to_csv('../processed/letterboxd_clean.csv', index=False)
print("FILE SAVING STATUS")
print(f"\n✓ Data save: ../processed/letterboxd_clean.csv")
print(f"✓ Total film: {len(df):,}")
print(f"✓ Total colom: {len(df.columns)}")
print(f"\nData ready notebook 02_recommendation_system.ipynb")
```
